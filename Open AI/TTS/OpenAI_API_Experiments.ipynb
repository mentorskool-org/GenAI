````python
import pandas as pd
from datetime import datetime, timedelta

# Load the API_CALLS_LOGS file
def log_api_call(text_used_for_speech: str, succeed: bool = True):
  csv_logs_path = '/content/drive/My Drive/Gen AI/OpenAI/API_CALLS_LOGS.csv'
  try:
    df = pd.read_csv(csv_logs_path)
  except Exception as error:
    df = pd.DataFrame()

  temp_df = pd.DataFrame([{
        "Timestamp": (datetime.now() + timedelta(hours=5, minutes=30)).strftime("%d-%m-%Y %H:%M:%S"),
        "Tokens Passed": len(tokenizer.encode(text_used_for_speech)),
        "Succeed": succeed
    }])
  df = pd.concat([df, temp_df]).reset_index(drop=True)
  df.to_csv(csv_logs_path, index=False)
````

### Generate conversational voice between Mentor and Mentee

````python
%pip install openpyxl
````

````python
mentor_and_mentee_conversations_df = pd.read_excel('/content/drive/My Drive/Gen AI/OpenAI/Mentor_Mentee_Conversation.xlsx')
mentor_and_mentee_conversations_df
````

````python
mentor_and_mentee_conversations_dict = mentor_and_mentee_conversations_df.to_dict(orient='records')
mentor_and_mentee_conversations_dict[:2]
````

````python
# Audio should get generated in the directory called conversational audio
for conversation in mentor_and_mentee_conversations_dict:
    file_name = f"conversation_{conversation['Seq']}_{conversation['Role']}.mp3"
    if conversation["Role"] == "Mentor":
        text_to_speech(
            conversation["Conversation"],
            voice="onyx",
            speech_file_path=f"conversational_audio/{file_name}",
            speed=0.9,
        )
    else:
        text_to_speech(
            conversation["Conversation"],
            voice="nova",
            speech_file_path=f"conversational_audio/{file_name}",
            speed=1,
        )
````

Now merge all generated voices to make a single conversation

````python
%pip install pydub
````

````python
from pydub import AudioSegment

# Initialize an empty AudioSegment
combined = AudioSegment.empty()

# Iterate through the files in sequence
for index, row in mentor_and_mentee_conversations_df.iterrows():
    role = row['Role']
    seq = row['Seq']
    filename = f'conversational_audio/conversation_{seq}_{role}.mp3'

    # Load the audio file
    audio = AudioSegment.from_mp3(filename)

    # Append to the combined AudioSegment
    combined += audio + AudioSegment.silent(duration=500)  # Add a brief pause between sentences

# Export the combined audio file
combined.export("combined_conversation.mp3", format="mp3")

print("Combined audio file has been generated as 'combined_conversation.mp3'.")
````
