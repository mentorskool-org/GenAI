````python
# Load the API_CALLS_LOGS file
def log_api_call(text_used_for_speech: str):
  file_path = '/content/drive/My Drive/Gen AI/OpenAI/API_CALLS_LOGS.txt'

  with open(file_path, 'r') as file:
      text = file.read()

  modified_text = text + "\nText: " + text_used_for_speech

  try:
    with open(file_path, 'w') as file:
        file.write(modified_text)
  except Exception as error:
    print(error)
    with open(file_path, 'w') as file:
        file.write(text)
````

````python
from pathlib import Path
from openai import OpenAI

client = OpenAI()
````

````python
def text_to_speech(text: str, voice:str = "alloy", speech_file_path: str = "speech.mp3", speed: int = 1.0):
  response = client.audio.speech.create(
    model="tts-1",
    voice=voice,
    input=text,
    speed=speed
  )

  response.stream_to_file(speech_file_path)

  # log the api call
  log_api_call(text)
````
