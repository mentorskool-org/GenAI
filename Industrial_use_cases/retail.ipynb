When we asked "How much is the price of the inventory for all small size t-shirts?". In this case LLM is thinking that

```
|t_shirt_id | brand  | color | size | price | stock_quantity |
| 8  | Adidas     | Black               | S                  | 14.0              | 89                |
```

Now in this case LLM thinks that Price column contains the total price of all the t_shirts. Like sum of 89 t_shirts of Adidas Black color of Small size is 14. Means (unit_price*89) = 14.

As column name is ot price per unit. Thus, it assumes that as total_price. This is the example that in real life column names won't be perfect.

````python
# Here we can also run the sql query
total_price = db_chain.run("SELECT SUM(price * stock_quantity) FROM t_shirts WHERE size = 'S'")
````

````python
# Let's ask complex query (includes joins)
query = """If we have to sell all the Adidas T-shirts today with discounts applied. How much revenue my store will generate."""
total_revenue = db_chain.run(query)
````

Here it is considering date also in discounts. Now based on LLM knowledge, discount should have date column as it would be between some dates and won't be permanent. But in our case there is no date column. Thus we have to tell LLM to only consider the schema of our database and based on that give the output

````python
# Let's run the sql query
sql_query = """
    select sum(a.total_amount * (1 - (coalesce(pct_discount, 0)/100.0)))
    as total_revenue from
    (select t_shirt_id, SUM(price*stock_quantity)as total_amount 
    from t_shirts ts
    where brand = 'Adidas' 
    group by t_shirt_id) a
    left join discounts d on
    a.t_shirt_id=d.t_shirt_id;
"""
qns5 = db_chain.run(sql_query)
````

**So let's do the Few-Shot learning to improve the response of the LLM**

### Few-shot Learning

````python
few_shots = [
    {
        "Question": "How many t-shirts do we have left for Nike in small size and white color?",
        "SQLQuery": "SELECT sum(stock_quantity) FROM t_shirts WHERE brand = 'Nike' AND color = 'White' AND size = 'S'",
        "SQLResult": "Result of the SQL query",
        "Answer": qns1,
    },
    {
        "Question": "How many t-shirts do we have left for Nike in extra small size and white color?",
        "SQLQuery": "SELECT sum(stock_quantity) FROM t_shirts WHERE brand = 'Nike' AND color = 'White' AND size = 'XS'",
        "SQLResult": "Result of the SQL query",
        "Answer": qns2,
    },
    {
        "Question": "How many t-shirts of Nike are left in the store?",
        "SQLQuery": "SELECT sum(stock_quantity) FROM t_shirts WHERE brand = 'Nike'",
        "SQLResult": "Result of the SQL query",
        "Answer": '517',
    },
    {
        "Question": "How much is the price of the inventory for all small size t-shirts?",
        "SQLQuery": "SELECT SUM(price*stock_quantity) FROM t_shirts WHERE size = 'S'",
        "SQLResult": "Result of the SQL query",
        "Answer": "20985",
    },
    {
        "Question": "If we have to sell all the Adidas T-shirts today with discounts applied. How much revenue my store will generate (post discounts)?",
        "SQLQuery": """SELECT sum(a.total_amount * ((100-COALESCE(discounts.pct_discount,0))/100)) as total_revenue from
        (select sum(price*stock_quantity) as total_amount, t_shirt_id from t_shirts where brand = 'Adidas'
        group by t_shirt_id) a left join discounts on a.t_shirt_id = discounts.t_shirt_id
        """,
        "SQLResult": "Result of the SQL query",
        "Answer": qns5,
    },
    {
        "Question": "If we have to sell all the Levi's T-shirts today. How much revenue our store will generate without discount?",
        "SQLQuery": "SELECT SUM(price * stock_quantity) FROM t_shirts WHERE brand = 'Levi'",
        "SQLResult": "Result of the SQL query",
        "Answer": "21473",
    },
    {
        "Question": "How many white color Levi's t-shirts we have available?",
        "SQLQuery": "SELECT sum(stock_quantity) FROM t_shirts WHERE brand = 'Levi' AND color = 'White'",
        "SQLResult": "Result of the SQL query",
        "Answer": "196",
    }
]
````

Creating Semantic Similarity Based example selector

- create embedding on the few_shots
- Store the embeddings in Chroma DB
- Retrieve the the top most Semantically close example from the vector store

Q.) Why we use the SemanticSimilarityExampleSelector

[Click here for reference](https://www.ateam-oracle.com/post/extending-saas-by-aimlpart-using-langchain-prompt-templates-few-shot-learning#:~:text=Use%20SemanticSimilarityExampleSelector%20to%20select%20prompt,and%20do%20a%20similarity%20search.)

Ans.) Use SemanticSimilarityExampleSelector to select prompt examples by similarity. It does this by finding the examples with the embeddings that have the greatest cosine similarity with the inputs. In this example, Chroma is used as the VectorStore class that is used to store the embeddings and do a similarity search. 

````python
%pip install huggingface_hub
````

````python
%pip install sentence-transformers
````

````python
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import Chroma

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
````

````python
# Here we are combining all the values of a dictionary. Like "<Question_value> <SQLQuery value> <SQLResult value> <Answer value>"
to_vectorize = [" ".join(example.values()) for example in few_shots]
````

````python
to_vectorize
````

This we are doing to vectorize this in the vector db, so that based on our prompt the most similar vector should be chosen out of all of these vectors via cosine similarity

[Datacamp Chroma Guide](https://www.datacamp.com/tutorial/chromadb-tutorial-step-by-step-guide)

````python
%pip install chromadb
````

````python
vectorstore = Chroma.from_texts(to_vectorize, embeddings, metadatas=few_shots)

# Have a doubt in this
````

````python
from langchain.prompts import SemanticSimilarityExampleSelector

# I created a example_selector that will fetch the 2 similar looking sentence (vector) based on our input (vector)
example_selector = SemanticSimilarityExampleSelector(
    vectorstore=vectorstore,
    k=2,
)
````

````python
example_selector.select_examples({"Question": "How many Adidas T shirts I have left in my store?"})
````

Now let's give some instruction to LLM, so Langchain provides the default prompt to us, we can use that, modify that and give it to LLM

````python
from langchain.chains.sql_database.prompt import PROMPT_SUFFIX, _postgres_prompt
````

````python
print(_postgres_prompt)
````

````python
print(PROMPT_SUFFIX)
````

Carefully, observe that the same format we have used in few-shot to adjust it based on our prompts


- **Question:** Question here
- **SQLQuery:** SQL Query to run
- **SQLResult:** Result of the SQLQuery
- **Answer:** Final answer here

````python
# Now based on prefix and postgre_sql prompt create a final prompt using prompt template
from langchain.prompts.prompt import PromptTemplate

example_prompt = PromptTemplate(
    input_variables=["Question", "SQLQuery", "SQLResult", "Answer"],
    template="\nQuestion: {Question}\nSQLQuery: {SQLQuery}\nSQLResult: {SQLResult}\nAnswer: {Answer}"
)
````

````python
from langchain.prompts import FewShotPromptTemplate

few_shot_prompt = FewShotPromptTemplate(
    example_selector=example_selector,
    example_prompt=example_prompt,
    prefix=_postgres_prompt,
    suffix=PROMPT_SUFFIX,
    input_variables=["input", "table_info", "top_k"]
)
````

````python
print(few_shot_prompt)
````

````python
db_chain_after_fewshot = SQLDatabaseChain.from_llm(
    llm,
    db,
    verbose=True,
    prompt=few_shot_prompt
) # the extra parameter that we pass is prompt to give the instructions. Now the response will be generated based on this
````

````python
db_chain_after_fewshot.run("How many white color Levi's shirt I have?")
````

````python
# Now let's ask the question of discount one, where we got the error when we tried before fewshot learning
query = """If we have to sell all the Adidas T-shirts today with discounts applied. How much revenue my store will generate."""
total_revenue = db_chain_after_fewshot.run(query)
````

````python
# Let's again run this with that exact prompt that we provided in the few_shot
query = """If we have to sell all the Adidas T-shirts today with discounts applied. How much revenue my store will generate. (post discounts)?"""
total_revenue = db_chain_after_fewshot.run(query)
````
