```python
from langchain.chat_models import ChatOpenAI
from langchain.vectorstores import DeepLake
from langchain.chains import RetrievalQA
from dotenv import load_dotenv, find_dotenv
# import deeplake
```

```python
load_dotenv(find_dotenv())
```

```python
from langchain.embeddings import OpenAIEmbeddings
```

```python
embeddings = OpenAIEmbeddings(model='text-embedding-ada-002')
```

```python
import os
```

```python
# Create a deep lake object and fetch information from deep lake
my_activeloop_org_id = os.environ.get("my_activeloop_org_id")
my_activeloop_datset_name = os.environ.get("my_activeloop_datset_name")
dataset_path = f"hub://{my_activeloop_org_id}/{my_activeloop_datset_name}"
```

```python
# Now load the dataset
vector_db = DeepLake(dataset_path=dataset_path, embedding=embeddings)
```

```python
# ds = deeplake.load(dataset_path)
# ds.visualize()
```

```python
retriever = vector_db.as_retriever()
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=ChatOpenAI(),
    chain_type="stuff",
    retriever=retriever
)
```

```python
response = qa_chain.run("List down all the table of enqurious etl db in list format.")
```
