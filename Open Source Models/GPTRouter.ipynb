```python
from dotenv import load_dotenv

load_dotenv()
```

```python
import os

api_key=os.environ.get("GPT_ROUTER_API_KEY")
```

```python
from gpt_router.client import GPTRouterClient
from gpt_router.models import ModelGenerationRequest, GenerationParams
from gpt_router.enums import ModelsEnum, ProvidersEnum
from gpt_router.constants import DEFAULT_API_BASE_URL


client = GPTRouterClient(base_url=DEFAULT_API_BASE_URL, api_key=api_key)
client
```

```python
messages = [
    {"role": "user", "content": "Write me a short poem"},
]
prompt_params = GenerationParams(messages=messages)
claude2_request = ModelGenerationRequest(
    model_name=ModelsEnum.CLAUDE_INSTANT_12,
    provider_name=ProvidersEnum.ANTHROPIC.value,
    order=1,
    prompt_params=prompt_params,
)

response = client.generate(ordered_generation_requests=[claude2_request])
print(response.choices[0].text)
```

```python
def generate_text(query):
    messages = [
        {"role": "user", "content": query},
    ]
    prompt_params = GenerationParams(messages=messages)
    claude2_request = ModelGenerationRequest(
        model_name=ModelsEnum.CLAUDE_INSTANT_12,
        provider_name=ProvidersEnum.ANTHROPIC.value,
        order=1,
        prompt_params=prompt_params,
    )

    response = client.generate(ordered_generation_requests=[claude2_request])
    return response
```

```python
query = "Who won the icc champion's trophy 2013?"
response = generate_text(query)
```

```python
print(response.choices[0].text)
```
