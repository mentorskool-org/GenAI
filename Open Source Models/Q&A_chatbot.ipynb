```python
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.llms import HuggingFaceHub
from langchain.text_splitter import CharacterTextSplitter
from langchain.vectorstores import Chroma
from langchain.chains import RetrievalQA
from langchain.document_loaders import PyPDFLoader
from dotenv import load_dotenv
import os



load_dotenv()
```

```python
# Use the falcon model
repo_id = "mistralai/Mistral-7B-v0.1"

# Create the llm
llm = HuggingFaceHub(
    huggingfacehub_api_token=os.environ["HUGGINGFACEHUB_API_KEY"],
    repo_id=repo_id,
    model_kwargs={"temperature": 0.1, "max_new_tokens": 500}
)
```

```python
response = llm.invoke("Who is MS Dhoni")
```

```python
print(response)
```

```python
# Let's load the pdf file
loader = PyPDFLoader("HighPerformanceAI.pdf")
```

```python
pages = loader.load_and_split() # using this pdf is page wise divided. It returns the
# list where each element contains the content of a page
```

```python
# Chunk the data, create embeddings, and save in vectorstore
text_splitter = CharacterTextSplitter(chunk_size=500, chunk_overlap=5)
docs = text_splitter.split_documents(pages)
# CharacterTextSplitter will only split on separator (which is '\n\n' by default).
# chunk_size is the maximum chunk size that will be split if splitting is possible.
```

```python
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
# Equivalent to SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")

# we have install the sentence-transformers package
```

```python
vectorstore = Chroma.from_documents(docs, embeddings)
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm, retriever=vectorstore.as_retriever(), chain_type="stuff"
)
```

```python
response = qa_chain.run("Organizations’ top data priorities over the next two years fall into three areas, all supported by wider adoption of cloud platforms:") 
```

```python
print(response)
```
